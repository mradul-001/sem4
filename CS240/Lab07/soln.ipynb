```python
import numpy as np
import pandas as pd
import threading
from sklearn import svm
from sklearn.metrics import f1_score, precision_score, recall_score
```

```python
train_data=pd.read_csv(r"./mnist_train.csv")
test_data=pd.read_csv(r"./mnist_test.csv")

train_data=train_data.to_numpy()    # train_data shape: (60000, 785)
m, n = train_data.shape
test_data=test_data.to_numpy()      # test_data shape: (10000, 785)

X_train=train_data[:,1:]            # X_train shape: (60000, 784)
y_train=train_data[:,0]             # y_train shape: (60000,)
X_test=test_data[:,1:]              # X_test shape: (10000, 784)
y_test=test_data[:,0]               # y_test shape: (10000,)

X_train = X_train / 255.0           # Normalizing the data
X_test = X_test / 255.0
```

### (a) Binary classifier to distinguish the digit ‘0‘ from all other digits

```python
y_tr = [1 if ele == 0 else 0 for ele in y_train]
y_te = [1 if ele == 0 else 0 for ele in y_test]
```

```python
clf = svm.SVC()
clf.fit(X_train[:], y_tr[:])
```

```python
y_pred = clf.predict(X_test)
```

```python
f1 = f1_score(np.array(y_te), y_pred)
pre = precision_score(y_te, y_pred)
rec = recall_score(y_te, y_pred)
```

```python
print("Scores for 0 classifier:")
print("F1 Score".ljust(20) + ":" + str(f1))
print("Precision Score".ljust(20)  + ":" + str(pre))
print("Recall Score".ljust(20)  + ":" + str(rec))
print()
print()
```

### (b) Training the classifier for all digits

```python
f1_scores = []
pr_scores = []
re_scores = []
```

```python
def classifier(x_train, y_train, x_test, y_test, target):

    print("entered the function")
    
    y_tr = [1 if ele == target else 0 for ele in y_train]
    y_te = [1 if ele == target else 0 for ele in y_test]

    model = svm.SVC()
    model.fit(x_train[:], y_tr[:])

    print("Fitted the mode for target: ", target)
    
    y_pred = model.predict(x_test)

    f1 = f1_score(np.array(y_te), y_pred)
    pre = precision_score(y_te, y_pred)
    rec = recall_score(y_te, y_pred)

    print()
    print("Classifier trained:")
    print("F1-Score =", f1)
    print("Precision Score =", pre)
    print("Recall score =", rec)
    print()

    f1_scores.append(f1)
    pr_scores.append(pre)
    re_scores.append(rec)

    return
```

#### Using multithreading for faster results

```python
threads = []

for i in range(10):
    thread = threading.Thread(target = classifier, args = [X_train, y_train, X_test, y_test, i])
    threads.append(thread)
    thread.start()

for i in range(10):
    threads[i].join()
```

```python
def print_results():
    print()
    print("Digit".ljust(20) + "F1 Score".ljust(25) + "Precision Score".ljust(25) + "Recall Score")
    print()
    for i in range(10):
        print(str(i).ljust(20) + str(round(f1_scores[i], 3)).ljust(25) + str(round(pr_scores[i], 3)).ljust(25)  + str(round(re_scores[i], 3)))
    print()
    
print_results()
```

- #### The digit 7 is the one with least scores, so we claim that it is the most difficult digit to classify.
